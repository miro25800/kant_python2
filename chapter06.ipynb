# 숫자 맞추기 게임

```python
import random # 난수 발생을 위한 모듈 임포트

# 1 ~ 100 사이의 난수 발생
number = random.randint(1, 100)

# 시도 횟수
try_count = 0

# 타이틀
print("="*45)
print("컴퓨터가 1부터 100사이의 숫자를 생성했습니다.")
print("="*45)


# 메인 루프
while True:
    user_input = input("컴퓨터의 숫자를 맞춰보세요. (1 ~ 100): ")

    if user_input == "q":
        print(f"게임 종료합니다. 정답은 {number}입니다.")
        break

    if not user_input.isdecimal():
        print("숫자만 입력해주세요.")
        continue

    user_number = int(user_input)
    # 시도 회수 증가
    try_count += 1 # try_count++ , ++try_count

    # 입력값과 정답을 비교
    if user_number < number:
        print("Up! 더 큰 수입니다.")
    elif user_number > number:
        print("Down! 더 작은 수 입니다.")
    else:
        print(f"정답입니다. {try_count}회 만에 성공했습니다.")
        break # 정답일 경우 무한 루프 탈출
```
