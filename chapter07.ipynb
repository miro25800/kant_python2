## 함수 (Function)

- 반복되는 로직을 합쳐놓은 문법
- 재사용 높일수 있는 코드 기법 # 프로그램 내 내장된 기법
- 단독사용가능 print(), len(), ...
- 함수명 : snake_case
- 매개 변수 (Parameter)

```python
def 함수이름(매개변수, 매개변수, ...):
    실행할 로직들 #1
    실행할 로직들 #2
    실행할 로직들 #3
```

````python

# 함수 정의
# '='같은 디폴트 값은 맨 뒤에 적어야 하며, 앞에 적을 수 없다. 
def say_hello(name, age, point=1000):
    print(f"안녕하세요. 저는 {name}입니다. 나이는 {age}입니다.")
    print(f"초기 지급 포인트 {point}가 지급되었습니다.")

say_hello("양건", 36, 2000)
````

````python
# 가변 인수
# *args

def add_all(*numbers): # 매개 변수 (Parameter)
    print(f"입력된 튜플: {numbers}")
    total = sum(numbers)
    print(f"합계 결과: {total}") # 여기까지는 정의이다.

add_all(1, 2, 3) # 인수(argument) # 모든 수를 합한다.
add_all(1, 2, 3, 4, 5)
add_all()

````

````python
# 딕셔너리 가변 인수
# **kwargs

def user_info(**data):
    print(f"전달된 딕셔너리 구조: {data}")
    for key, value in data.items():
        print(f"{key}: {vlaue}")

# 함수 호출
user_info(name="스파르타", level=5, active=True)
````

## 변수 스코프 (유효범위)

- 전역 변수 : 함수 밖에서 선언하는 변수, 어디서나 접근이 가능
- 지역 변수 : 함수내에 선언하는 변수, 함수안에서만 접근이 가능 # 함수가 소멸되면 자동적으로 소멸된다.

````python
def local():
    local_var = 1 # 지역변수 # 지역변수는 함수안에서만 영향을 미친다.
    print(local_var)

local()
# print(local_var) # 오류 발생
````

````python
# 전역 변수
message = "안녕하세요. kant몰에 오신것을 환영합니다."

def greeting():
    message = "장남감 코너는 3층에 있습니다." # 지역 변수 # 지역 변수가 전역 변수보다 우선 출력된다.
    print(message) # 지역 변수가 우선된다.

greeting()

print(message) # 전역 변수
````

````python
# 펑션 안에 펑션을 넣는다.
# 전역 변수
value = "전역 데이터"

# Function =================
def outer_function():
    value = "지역 변수 1"

    def inner_function():
        value = "지역 변수 2"
        print(value)

    inner_function()
    print(value)
# Function =================

outer_function()
print(value) # 전역변수
````

````python
# global 키워드

counter = 0

def inc_counter():
    counter = 0
    # global count # 전역 변수를 참조하겠다 선언
    counter += 1

inc_counter()
inc_counter()

print(counter)
````

````python
# 일반함수
def add_five(x):
    return x + 5

print(add_five(10))

# 람다식
result = lambda x: x + 5
print(result(20))
````

````python
# map 모든 요소를 반환
numbers = [1, 2, 3, 4, 5]

result = list(map(lambda x: x + 10, numbers))
print(f"계산 결과:  {result}")

# 리스트 컴프리핸션
print([x+10 for x in numbers])

filter_result = list(filter(lambda x: x % 2 == 0, numbers))
print(f"짝수 결과: {filter_result}")

print([x for x in numbers if x % 2 == 0])
````

````python
# 정렬
players = [
    ("손흥민", 95),
    ("김민재", 60),
    ("이간인", 80)
]

print(sorted(players, key=lambda p: p[1])) # 오름차순
print(sorted(players, key=lambda p: p[1], reverse=True)) # 내림차순
print(max(players, key=lambda p: p[1])) # 최고 점수 추출
````

````python
numbers = [2, 3, 4, 5, 6, 7]

# Lambda 식으로 모든 요소의 거듭제곱 추출
sqr_numbers = list(map(lambda n : n ** 2, numbers))  # 이것은 일종의 함수이다.
print(f"거듭제곱 결과값: {sqr_numbers}")

# 결과값이 20 이상인 값만 추출
sqr_number_over_20 = list(filter(lambda n : n > 20, sqr_numbers))
print(f"거듭제곱값이 20 이상: {sqr_number_over_20}")
````
